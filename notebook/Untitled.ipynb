```python
import numpy as np
import uproot
import datetime

# Dummy input
filenames = ["dummy1.h5", "dummy2.h5"]
channels = [0, 1, 2]

# Number of events to simulate
n_events = 1000000
n_channels = len(channels)

# Output ROOT file
with uproot.recreate("output3.root") as rootfile:
    # Define tree structure
    tree_data = {
        "eventID": np.int32,
        "timestamp": np.uint64,
        "timetag": np.uint32,
        "baseline_mean": "float32",
        "baseline_std": "float32",
        "t_max": "float32",
        "pulse_height": "float32",
        "risetime": "float32",
        "falltime": "float32",
        "pulse_integral": "float32",
        "pulse_width": "float32",
        "tail_to_total": "float32",
        "psd_param": "float32",
        "famr": "float32",
        "saturation_flag": "bool",
        "pileup_flag": "bool",
        "valid_flag": "bool",
        "channel": np.int16,
        "filename": str,
    }

    tree = rootfile.mktree("Events", tree_data)

    # Simulate and write event-by-event
    for event_id in range(n_events):
        
        print( "Processed {:.2f}".format(100*event_id/n_events), end='\r' )
        
        timestamp = int(datetime.datetime.now().timestamp())
        timetag = np.random.randint(0, 1e6)

        for ch in channels:
            # Simulate waveform data (not used in this example)
            waveform = np.random.normal(0, 1, 100)

            # Dummy processed quantities
            ch_data = {
                "eventID": event_id,
                "timestamp": timestamp,
                "timetag": timetag,
                "baseline_mean": np.random.uniform(0, 1),
                "baseline_std": np.random.uniform(0, 0.2),
                "t_max": np.random.uniform(10, 50),
                "pulse_height": np.random.uniform(5, 10),
                "risetime": np.random.uniform(1, 5),
                "falltime": np.random.uniform(5, 10),
                "pulse_integral": np.random.uniform(50, 100),
                "pulse_width": np.random.uniform(8, 12),
                "tail_to_total": np.random.uniform(0.1, 0.5),
                "psd_param": np.random.uniform(0, 1),
                "famr": np.random.uniform(0, 1),
                "saturation_flag": np.random.choice([True, False]),
                "pileup_flag": np.random.choice([True, False]),
                "valid_flag": True,
                "channel": ch,
                "filename": filenames[event_id % len(filenames)],
            }

            # Extend the tree with one entry
            tree.extend({k: [v] for k, v in ch_data.items()})
```
