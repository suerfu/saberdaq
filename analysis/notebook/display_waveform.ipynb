```python
x1 = np.linspace(-10,10,100)
y1 = np.exp(-x1)
y1[x1<0] = 0

x2 = np.linspace(-10,10,200)
y2 = np.exp(-x2)
y2[x2<0] = 0

plt.plot( x1, y1 )
plt.plot( x2, y2 )
#plt.yscale( 'log' )
```

```python
F1 = np.fft.rfftfreq( len(x1), x1[1]-x1[0] )

Y1 = np.fft.rfft(y1)
Y1 /= (F1[1]-F1[0]) * len(F1)

F2 = np.fft.rfftfreq( len(x2), x2[1]-x2[0] )

Y2 = np.fft.rfft(y2)
Y2 /= (F2[1]-F2[0]) * len(F2)

plt.plot( F1, np.abs(Y1) )
plt.plot( F2, np.abs(Y2) )
plt.yscale( 'log' )
```
