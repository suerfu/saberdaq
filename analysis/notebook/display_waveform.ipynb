```python
x = np.linspace(-10,10,100)
y = np.exp(-x)
y[x<0] = 0

plt.plot( x, y )
#plt.yscale( 'log' )
```

```python
F = np.fft.rfftfreq( len(x), x[1]-x[0] )

Y = np.fft.rfft(y)
Y /= (F[1]-F[0]) * len(x)

plt.plot( F, np.abs(Y) )
plt.yscale( 'log' )
```
